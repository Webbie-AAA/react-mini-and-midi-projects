# React 

### A JS library for building user interfaces

- Repeatable components
- Complex user interfaces become component trees, of resuable blocks of code, customised for different functions
- Creates a vastly simplified web structure, because code is reusable and modular. 
- Creates apps that scale really well
- React combines small amounts of HTML, CSS and JS into a single component. 
- Twitter uses React
- Each component listens to the server and updates itself and talk to server independently - without affecting any of the rest of the website. 
- Diffing: comparing changes. When a user makes a change, it compares the new version of the DOM tree that you want rendered, to the old tree, and then re-renders only the component that had the difference made to.

## Components

#### Splitting a complex web structure into small reusable pieces

##### How To Create A Custom Component

Custom components are written in title case so they're

not confused with HTML tags.

```python
IN JSX:
function Name(){
    return <h3>Umelbaneen Al-Mousawi</h3>
}

function Nickname(){
    return <h3>Umel</h3>
}

In React App:
<Name />
<Nickname />
```

#### Convention is to leave our App.jsx near  empty, with only the < App /> rendered and write components in separate JSX files

Then export and import between files to use the components.

### Deep Dive On Imports & Exports

One theres only __one__ possible thing to export:

        export default <function_name>

When we __default export__, in the import, we can import it with any name

We export __multiple___ things, we use curly brackets:

        export {thing_1, thing_2, thing_3 .... thing_n}

We need to import multiple things like so:

        import {thing_1, thing_2, thing_3 ... thing_n} from file_path

These need to be accurately named

We can export a default function

And below it export multiple functions

That was a module can have a default return, but with other options available if called

`import * as alias from file_path`

can be called as

`alias.default` for the default export
`alias.thing1` `alias.thing2`, and so on for the other exports.

# Prop

Attributes that send custom pieces of data to our component

for example, input property has an fname='' attribute


In the same way, in our custom components, we can provide attributes, and they're called __props__.

```python
function Card(props){

    return(
        <h2>My name is {props.name}</h2>
    )
}

//send over the property as:
<Card name='Benoon'/>
```
